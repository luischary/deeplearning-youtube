```python
import numpy as np
import matplotlib.pyplot as plt

TRUE_W = 2.0
TRUE_B = 3.0
```

```python
N_SAMPLES = 100
NOISE_STD = 0.5
SEED = 42

rng = np.random.default_rng(SEED)
x = rng.uniform(-2.0, 2.0, size=N_SAMPLES)
y = TRUE_W * x + TRUE_B + rng.normal(0.0, NOISE_STD, size=N_SAMPLES)
```

```python
print(x.shape)
print(y.shape)
```

```python
x[:5]
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(x, y, s=35, alpha=0.75, label="dados")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Dados sintéticos")
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```
